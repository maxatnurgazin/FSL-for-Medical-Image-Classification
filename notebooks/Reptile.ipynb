```python
import torch
import numpy as np
def mixup(data, target, alpha):
    # Get the batch size and generate the beta distribution
    batch_size = data.size(0)
    lam = np.random.beta(alpha, alpha)

    # Create random indices for the mix-up
    index = torch.randperm(batch_size).to(data.device)

    # Apply MixUp
    mixed_data = lam * data + (1 - lam) * data[index]
    target_a, target_b = target, target[index]

    # Return the mixed data and mixed labels
    return mixed_data, target_a, target_b, lam

def cutmix(data, target, alpha):
    # Get the batch size and generate the beta distribution
    batch_size = data.size(0)
    # lam = np.random.beta(alpha, alpha)
    lam = 0.5

    # Create random indices for the mix-up
    index = torch.randperm(batch_size).to(data.device)

    # Calculate the bounding box size
    bbx1, bby1, bbx2, bby2 = rand_bbox(data.size(), lam)

    # Apply CutMix
    data[:, :, bbx1:bbx2, bby1:bby2] = data[index, :, bbx1:bbx2, bby1:bby2]
    target_a, target_b = target, target[index]

    # Return the mixed data and mixed labels
    return data, target_a, target_b, lam

def cutout(data, size, num_holes):
    # Get the data dimensions
    _, _, H, W = data.size()

    # Create a mask
    mask = np.ones((H, W), dtype=np.float32)

    # Apply cutout
    for _ in range(num_holes):
        y = np.random.randint(H)
        x = np.random.randint(W)
        y1 = np.clip(y - size // 2, 0, H)
        y2 = np.clip(y + size // 2, 0, H)
        x1 = np.clip(x - size // 2, 0, W)
        x2 = np.clip(x + size // 2, 0, W)
        mask[y1:y2, x1:x2] = 0

    # Apply mask to the data
    data = data * torch.from_numpy(mask).to(data.device)
    return data

def rand_bbox(size, lam):
    W = size[2]
    H = size[3]
    cut_rat = np.sqrt(1. - lam)
    cut_w = np.int(W * cut_rat)
    cut_h = np.int(H * cut_rat)

    # uniform
    cx = np.random.randint(W)
    cy = np.random.randint(H)

    bbx1 = np.clip(cx - cut_w // 2, 0, W)
    bby1 = np.clip(cy - cut_h // 2, 0, H)
    bbx2 = np.clip(cx + cut_w // 2, 0, W)
    bby2 = np.clip(cy + cut_h // 2, 0, H)

    return bbx1, bby1, bbx2, bby2
```

```python
iters = [5]
for ite in iters:
    train_dataset_iterator  = iter(train_loader)
    val_dataset_iterator    = iter(val_loader)
    
    num_inner_loop      = ite
    num_inner_loop_eval = ite

    model = create_model(model_name)
    criterion = nn.CrossEntropyLoss()

    n_epochs = 100
    scheduler_milestones = [120, 160]
    scheduler_gamma = 0.1
    learning_rate = 1e-5
    tb_logs_dir = Path(".")

    train_optimizer = SGD(
        model.parameters(), 
        lr=learning_rate, 
        momentum=0.9, 
        weight_decay=5e-4,
    )

    warmup_scheduler = LinearLR(
        train_optimizer,
        start_factor=0.02,
        end_factor=1, 
        total_iters=5
    )
    cosine_scheduler = CosineAnnealingLR(
        train_optimizer,
        T_max=n_epochs-5,
        eta_min=1e-6
    )

    seq_scheduler = SequentialLR(
        train_optimizer, 
        [warmup_scheduler, cosine_scheduler],
        milestones=[5]
    )


    train_scheduler = MultiStepLR(
        train_optimizer,
        milestones=scheduler_milestones,
        gamma=scheduler_gamma,
    )

    train_scheduler = seq_scheduler
    if tb_write:
        writer = SummaryWriter(comment=f'reptile_{model_name}_{dataset}_inner{num_inner_loop}')
    with tqdm(range(num_meta_iterations), position=0, desc="Outer loop", leave=True) as reptile_tqdm:
        for meta_iter in reptile_tqdm:
            model.train()
            
            task_models = []

            for _ in range(train_batch_size):
                support_images, support_labels, query_images, query_labels, _ = next(train_dataset_iterator)
            
                task_model = copy.deepcopy(model).cuda()
                task_optimizer = optim.SGD(task_model.parameters(), lr=inner_loop_lr)

                for inner_iter in range(num_inner_loop):
                    data, target = support_images.to(DEVICE), support_labels.to(DEVICE)
                    
                    task_optimizer.zero_grad()

                    if augmentation == 'cutmix':
                        data, target_a, target_b, lam = cutmix(data, target, alpha=1.0)
                        target = lam * target_a + (1 - lam) * target_b
                    elif augmentation == 'mixup':
                        data, target_a, target_b, lam = mixup(data, target, alpha=1.0)
                        target = lam * target_a + (1 - lam) * target_b
                    elif augmentation == 'cutout':
                        data = cutout(data, size=64, num_holes=1)
                    preds = task_model(data)

                    if augmentation == 'cutmix' or augmentation == 'mixup':
                        loss = lam * criterion(preds, target_a) + (1 - lam) * criterion(preds, target_b)
                    else:
                        loss = criterion(preds, target)
                    # loss = criterion(preds, target)
                    
                    loss.backward()
                    
                    task_optimizer.step()

                task_models.append(task_model.cpu())

            for model_param, *task_model_params in zip(model.parameters(), *(task_model.parameters() for task_model in task_models)):
                model_param.data = model_param.data + (meta_step_size * (sum(task_model_param.data for task_model_param in task_model_params) / train_batch_size - model_param.data))

            
            # for task_model in task_models:
            #     del task_model
            task_models = []
            # Evaluation
            if meta_iter % 5 == 100:
                model.eval()
                total_accuracy = 0
                num_eval_tasks = eval_batch_size

                for _ in range(eval_batch_size):
                    support_images, support_labels, query_images, query_labels, _ = next(val_dataset_iterator)
                    task_model = copy.deepcopy(model).cuda()
                    task_optimizer = optim.SGD(task_model.parameters(), lr=inner_loop_lr)

                    for inner_iter in range(num_inner_loop_eval):
                        data, target = support_images.to(DEVICE), support_labels.to(DEVICE)

                        task_optimizer.zero_grad()

                        preds = task_model(data)

                        loss = criterion(preds, target)
                        loss.backward()

                        task_optimizer.step()

                    # Test the adapted model on the query set
                    data, target = query_images.to(DEVICE), query_labels.to(DEVICE)
                    preds = task_model(data)
                    pred_labels = preds.argmax(dim=1)
                    accuracy = (pred_labels == target).float().mean().item()
                    total_accuracy += accuracy
                mean_accuracy = total_accuracy / num_eval_tasks
                if tb_write:
                    writer.add_scalar("acc", mean_accuracy, meta_iter)
                print(f"Meta-Iteration {meta_iter}: Validation accuracy: {mean_accuracy:.4f}")
                # reptile_tqdm.set_postfix(test_accuracy=mean_accuracy)
    
    
    test_batch_size = 100
    n_shots = [3, 5, 10]


    for number_shot in n_shots:
        test_sampler = TaskSampler(
            val_set, n_way=n_way, n_shot=number_shot, n_query=n_query, n_tasks=test_batch_size
        )

        test_loader = DataLoader(
            val_set,
            batch_sampler=test_sampler,
            num_workers=n_workers,
            pin_memory=True,
            collate_fn=test_sampler.episodic_collate_fn,
        )

        model.eval()
        total_accuracy = 0
        num_eval_tasks = len(test_loader)

        with tqdm(
                enumerate(test_loader), total=len(test_loader), desc="eval loop", position=1, leave=False,
            ) as tqdm_val:
                for episode_index, (
                    support_images,
                    support_labels,
                    query_images,
                    query_labels,
                    _,
                ) in tqdm_val:
                    # mem()

                    task_model = copy.deepcopy(model).cuda()
                    task_optimizer = optim.SGD(task_model.parameters(), lr=inner_loop_lr)

                    for inner_iter in range(num_inner_loop_eval):
                        data, target = support_images.to(DEVICE), support_labels.to(DEVICE)

                        task_optimizer.zero_grad()
                        
                        preds = task_model(data)

                        loss = criterion(preds, target)
                        loss.backward()

                        task_optimizer.step()

                    # Test the adapted model on the query set
                    data, target = query_images.to(DEVICE), query_labels.to(DEVICE)
                    preds = task_model(data)
                    pred_labels = preds.argmax(dim=1)
                    accuracy = (pred_labels == target).float().mean().item()
                    total_accuracy += accuracy
                    del preds

        mean_accuracy = total_accuracy / num_eval_tasks
        print(f"{model_name}-inner{num_inner_loop_eval}_{n_way}-way {number_shot}-shot Test accuracy: {mean_accuracy:.4f}")
```
